# Create Paper Figures

Runs all figure-generating notebooks using papermill. Each cell executes one notebook and saves
the executed output to `_paper_figures_output/`.

| Notebook | Paper figures |
|---|---|
| `_2-1 results person-attribution experiments` | A1, A2, A3, A4, A5 |
| `_2-2 plot asymmetries by word pairs` | Fig A6 |
| `_2-3 plot model position bias` | Fig A7, A8, A9 |
| `_3-1 results politicized-context experiments` | A10, A11 |
| `_4-0 create combined scatter plots` | Fig 1 (LM Arena), Fig A12 (ECI) |
| `_4-2 create bias ranking new` | Table 1 (`bias_ranking_table.tex`) |
| `_6-1 plot debiasing system prompts` | Fig 3 |
| `_6-2 plot reasoning effort experiments` | Fig 4 |
| `_6-3 plot thinking effort experiments grok` | Fig A13, A14 |
| `_7-1 plot explicit exclude prompts_strict results` | Fig 5 |

```python
import os
import papermill as pm

NOTEBOOKS_DIR = os.path.abspath(os.getcwd())
OUTPUT_DIR = os.path.join(NOTEBOOKS_DIR, '_paper_figures_output')
os.makedirs(OUTPUT_DIR, exist_ok=True)

def run_notebook(name):
    input_path = os.path.join(NOTEBOOKS_DIR, name)
    output_path = os.path.join(OUTPUT_DIR, name)
    pm.execute_notebook(
        input_path=input_path,
        output_path=output_path,
        cwd=NOTEBOOKS_DIR,
        kernel_name='python3',
        progress_bar=False,
    )
    print(f'Done: {name}')
```

```python
# Fig 1, A1 (heatmap), A2 (heatmap summary), A3 (ECI scatter), A4, A5 (subjective-only scatter)
run_notebook('_2-1 results person-attribution experiments.ipynb')
```

```python
# Fig A6 — model selection frequencies across left/right descriptor pairs
run_notebook('_2-2 plot asymmetries by word pairs.ipynb')
```

```python
# Fig A7 (signed position bias), A8 (position bias vs LM Arena), A9 (position bias vs ECI)
run_notebook('_2-3 plot model position bias.ipynb')
```

```python
# Fig 2 (LM Arena scatter), A10 (heatmap), A11 (ECI scatter)
run_notebook('_3-1 results politicized-context experiments.ipynb')
```

```python
# Combined 1×2 scatter plots (person-attribution + politicized-context) vs LM Arena and ECI
run_notebook('_4-0 create combined scatter plots.ipynb')
```

```python
# Table 1 — bias_ranking_table.tex
run_notebook('_4-2 create bias ranking new.ipynb')
```

```python
# Fig 3 — per-model bias distribution under three system-prompt conditions
run_notebook('_6-1 plot debiasing system prompts.ipynb')
```

```python
# Fig 4 — politicized-context bias as a function of reasoning effort
run_notebook('_6-2 plot reasoning effort experiments.ipynb')
```

```python
# Fig A12 (signed log-odds Grok reasoning), A13 (absolute log-odds Grok reasoning)
run_notebook('_6-3 plot thinking effort experiments grok.ipynb')
```

```python
# Fig 5 — per-model bias under base prompt vs explicit-exclusion (prompts_strict) condition
run_notebook('_7-1 plot explicit exclude prompts_strict results.ipynb')
```
