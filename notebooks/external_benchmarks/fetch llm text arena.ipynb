```python
import requests
import pandas as pd
#  overall rankings across various AI models in text-to-text tasks across math, coding, creative writing, and other open-ended domains.
dfs = []
for index in range(0, 300, 100):
    print(f"Fetching rows {index} to {index + 100}...")
    response = requests.get(f"https://datasets-server.huggingface.co/rows?dataset=lmarena-ai%2Fleaderboard-dataset&config=text_style_control&split=latest&offset={index}&length=100")
    df = pd.DataFrame([item['row'] for item in response.json()['rows']])
    dfs.append(df)
df = pd.concat(dfs, ignore_index=True)
df
```

```python
df = df[["model_name","rating"]]
df.to_csv("llm_arena_ratings.csv", index=False)
```
