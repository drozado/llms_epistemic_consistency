```python
from math import exp, log
from matplotlib.pyplot import plot
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import math
import scipy.stats as stats
from vpei.epistemic_consistency.results_utils import compute_models_overall_bias_in_politicized_context_experiments, DEFAULT_EXPERIMENTAL_RESULTS_PATH, compute_stats_from_experimental_results, load_models_experiment_results, compute_statistics_for_absolute_experiments, compute_statistics_for_comparative_experiments
from vpei.common_variables import EXPERIMENTS_WEIGHTS_FOR_OVERALL_BIAS_RATING, POLITICAL_POLES_PALETTE
from vpei.epistemic_consistency.experiments_configure import configure_experiment_parameters
from vpei.epistemic_consistency.results_utils import compute_models_overall_bias_ratings


experiments_types_and_names_to_load = {
    "unblind_experiment": ["evaluate_research_designs","evaluate_time_series_trends","evaluate_governments_based_on_country_metrics","evaluate_factuality_of_news_articles","evaluate_policy_proposals","evaluate_two_group_comparison_policy_effectiveness","evaluate_correlation_btw_governments_and_problem_metrics","evaluate_protesters_behavior","evaluate_social_media_posts","evaluate_policy_effectiveness_given_contingency_tables"],
}

target_statistic = 'log_odds'
# target_statistic = 'absolute_log_odds'
experiments_weights = EXPERIMENTS_WEIGHTS_FOR_OVERALL_BIAS_RATING

models = [
    "grok-4.20-non-reasoning",
    "grok-4.20-reasoning",
    "grok-4-1-fast-non-reasoning",    
    "grok-4-1-fast-reasoning",
]

df = compute_models_overall_bias_in_politicized_context_experiments(
    models,
    experiments_types_and_names_to_load,
    experimental_results_path=DEFAULT_EXPERIMENTAL_RESULTS_PATH,
    target_statistic=target_statistic,
)

d = {}
for i, row in df.iterrows():
    if 'non-reasoning' in row["model_name"]:
        if "non-reasoning" not in d:
            d["non-reasoning"] = {}
        d["non-reasoning"][row["model_name"].replace("-non-reasoning", "")] = row["MODEL MEAN"]
    else:
        if "reasoning" not in d:
            d["reasoning"] = {}
        d["reasoning"][row["model_name"].replace("-reasoning", "")] = row["MODEL MEAN"]
df = pd.DataFrame(d)
df


# create a plot where non-reasoning and reasoning are the only two values in the x axis and the y axis is the overall bias rating
# each model is shown as a marker, with lines connecting the same model across the two conditions
fig, ax = plt.subplots(figsize=(10, 6))

x_positions = {'non-reasoning': 0, 'reasoning': 1}
colors = plt.rcParams['axes.prop_cycle'].by_key()['color']
model_colors = {
    "grok-4.20": colors[0],
    "grok-4-1-fast": colors[1],
}

for i, model in enumerate(df.index):
    y_vals = [df.loc[model, 'non-reasoning'], df.loc[model, 'reasoning']]
    x_vals = [0, 1]
    color = model_colors.get(model, colors[i % len(colors)])
    ax.plot(x_vals, y_vals, marker='o', linestyle='--', markersize=8, label=model, color=color)

plt.axhline(0, color='gray', linestyle='--', linewidth=2)
ax.set_xticks([0, 1])
ax.set_xticklabels(['non-reasoning', 'reasoning'], fontsize=14)
ax.set_title('Overall Bias Rating for Reasoning vs Non-Reasoning Grok Models', fontsize=16)
ax.set_xlabel('Reasoning Condition')
ax.set_ylabel('Overall Bias Rating (Log Odds)', fontsize=14)
ax.legend(title='Model', fontsize=14)
plt.tight_layout()
fig.savefig('./figures/appendix_grok_reasoning.png', dpi=300, bbox_inches='tight')
plt.show()
```
