```python
thinking_levels = np.arange(df.shape[1])

model_correlations = {}
for row in df.iterrows():
    model_name = row[0]
    model_values = row[1].to_numpy().astype(float)
    corr = np.corrcoef(thinking_levels, model_values)[0,1]
    model_correlations[model_name] = corr
    print(f"Model: {model_name}, Correlation: {corr}")

#Fisher z-transformation
models_z_transformations = {}
for model_name, corr in model_correlations.items():
    z = np.arctanh(corr)
    models_z_transformations[model_name] = z
    print(f"Model: {model_name}, Fisher z-transformation: {z}")

z_mean = np.mean(list(models_z_transformations.values()))
z_transformed_mean_r = np.tanh(z_mean)
raw_mean_r = np.mean(list(model_correlations.values()))

print(f"raw mean correlation (r): {raw_mean_r}")
print(f"Overall z-transformed corrected mean correlation (r): {z_transformed_mean_r}")
                   
    
```
