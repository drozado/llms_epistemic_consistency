```python
import random
import json
import re
import os
import asyncio
import pandas as pd
import shutil
from datasets import load_dataset
from local_variables import PHYSICS_PROBLEMS_SUBJECTS
```

```python
dfs = []
for subject in PHYSICS_PROBLEMS_SUBJECTS:
    df = pd.read_csv(f"./data/{subject}.csv")
    dfs.append(df)
df = pd.concat(dfs)
df
```

```python
df_sample = df[(df['language']=="EN") & (df['answer_type']=="NV")].sample(2000, random_state=1)
output_file = "./data/sample_physics_problems_and_solutions.csv"
if not os.path.exists(os.path.dirname(output_file)):
    os.makedirs(os.path.dirname(output_file))
df_sample.to_csv(output_file, index=False)
```
